```python
import pandas as pd

DATA_DIR_PATH = "../data/"
```

```python
INPUT_PATH = DATA_DIR_PATH + "raw/scrape_kompas_all.xlsx"

df = pd.read_excel(INPUT_PATH)

df.head()
```

```python
INPUT_CLEAN_PATH = DATA_DIR_PATH + "raw/scrape_kompas_clean.csv"

df_clean = pd.read_csv(INPUT_CLEAN_PATH)

df_clean.head()
```

# Find the difference between clean and main Dataset Files

```python
# Find the difference between clean and main Dataset Files
diff = df_clean.merge(df, on="link", how="outer", indicator=True)

print("Difference:", diff["_merge"].value_counts())
diff.head()
```
